Array Functions

```python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# max/min/sum/prod
# 0 -> column and 1 -> rows
print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))
np.max(a1,axis = 1)
np.min(a1,axis=0)
```

```python
# mean/median/std/var
np.median(a1,axis=1)
np.mean(a1,axis=0)
np.std(a1,axis=0)
np.var(a1,axis=1)
```

```python
# trignomatric functions
np.sin(a1)
```

```python
# dot product
a2 = np.arange(12).reshape((3,4))
a3 = np.arange(12,24).reshape((4,3))
np.dot(a2,a3)
```

```python
# log and exponenets
np.log(a1)
np.exp(a1)
```

```python
# round/floor/ceil
print(np.round(np.random.random((3,4))*100))
print(np.floor(np.random.random((3,4))*100))
np.ceil((np.random.random((3,4))*100))
```
